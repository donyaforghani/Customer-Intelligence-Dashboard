```python
dt_clean= data1.dropna(subset=["Income"])
dt_clean= dt_clean[dt_clean["Income"]<150000]
dt_clean= dt_clean[dt_clean["Age"]<100]
```

```python
dt_clean["spend_segment"]= pd.qcut(dt_clean["total_spend"], q=3, labels=["low","medium","high"])
```

```python
fig, ax= plt.subplots()
dt_clean.boxplot(column="Income", by="spend_segment", ax=ax)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))

data_low = dt_clean[dt_clean['spend_segment'] == 'low']['Income']
data_med = dt_clean[dt_clean['spend_segment'] == 'medium']['Income']
data_high = dt_clean[dt_clean['spend_segment'] == 'high']['Income']
ax.boxplot([data_low, data_med, data_high], label=['Low', 'Medium', 'High'], patch_artist=True)

ax.set_title('Income Distribution by Spend Segment')
ax.set_xlabel('Spend Segment')
ax.set_ylabel('Income')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
scatter = ax.scatter(dt_clean['Income'], dt_clean['total_spend'],
                     c=dt_clean['Response'], cmap='coolwarm', alpha=0.5)
ax.set_title('Income vs Total Spend (colored by Campaign Response)')
ax.set_xlabel('Income')
ax.set_ylabel('Total Spend')
plt.colorbar(scatter, label='Response')
plt.show()
```

```python
category_totals = dt_clean[all_spend].sum().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(category_totals.index, category_totals.values, color='steelblue')
ax.set_title('Total Spending by Product Category')
ax.set_ylabel('Total Spend')
plt.xticks(rotation=45)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

# محاسبه میانگین سن برای هر گروه
age_means = dt_clean.groupby('spend_segment')['Age'].mean()

# رسم نمودار میله‌ای
ax.bar(age_means.index, age_means.values, color=['#A0C4FF', '#FFD166', '#EF476F'])
ax.set_title('Average Age by Spend Segment')
ax.set_xlabel('Spend Segment')
ax.set_ylabel('Average Age')

# نوشتن اعداد بالای میله‌ها
for i, v in enumerate(age_means.values):
    ax.text(i, v + 0.5, f"{v:.1f}", ha='center', va='bottom')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

# محاسبه میانگین تعداد فرزندان برای هر گروه
children_means = dt_clean.groupby('spend_segment')['children'].mean()

ax.bar(children_means.index, children_means.values, color=['#A0C4FF', '#FFD166', '#EF476F'])
ax.set_title('Average Number of Children by Spend Segment')
ax.set_xlabel('Spend Segment')
ax.set_ylabel('Average Children')

for i, v in enumerate(children_means.values):
    ax.text(i, v + 0.02, f"{v:.2f}", ha='center', va='bottom')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

# محاسبه میانگین پاسخ به کمپین‌ها (چون مقادیر ۰ و ۱ هستن، میانگین یعنی نرخ پاسخ)
cmp_cols = ['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5']
cmp_means = dt_clean.groupby('spend_segment')[cmp_cols].mean()

# رسم نمودار میله‌ای گروهی
cmp_means.plot(kind='bar', ax=ax, color=['#4C72B0', '#55A868', '#C44E52', '#8172B2', '#CCB974'])
ax.set_title('Campaign Acceptance Rate by Spend Segment')
ax.set_xlabel('Spend Segment')
ax.set_ylabel('Acceptance Rate (Mean)')
ax.set_xticklabels(cmp_means.index, rotation=0)
ax.legend(title='Campaign')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

scatter = ax.scatter(dt_clean['NumWebVisitsMonth'], dt_clean['total_spend'],
                     c=dt_clean['spend_segment'].map({'low': 0, 'medium': 1, 'high': 2}),
                     cmap='viridis', alpha=0.5)

ax.set_title('Web Visits vs Total Spend')
ax.set_xlabel('Monthly Web Visits')
ax.set_ylabel('Total Spend')

plt.show()
```

```python
# انتخاب فیچرهای کلیدی برای پرسونا
persona_features = ['Income', 'Age', 'children', 'total_spend', 'NumWebPurchases']
high_means = dt_clean[dt_clean['spend_segment'] == 'high'][persona_features].mean()
overall_means = dt_clean[persona_features].mean()

# نرمال‌سازی برای اینکه مقیاس‌ها یکی بشن (تقسیم بر میانگین کل)
normalized_high = high_means / overall_means

fig, ax = plt.subplots(figsize=(8, 5))
y_pos = np.arange(len(persona_features))

ax.barh(y_pos, normalized_high.values, color='#EF476F', alpha=0.8)
ax.axvline(1.0, color='black', linestyle='--', label='Overall Average')

ax.set_yticks(y_pos)
ax.set_yticklabels(persona_features)
ax.set_xlabel('Ratio compared to Overall Average')
ax.set_title('High-Value Customer Persona Profile')
ax.legend()

plt.show()
```
